```python
# Cvičení (spojení všeho): Průměr známek z libovolného počtu vstupů.
# Ptej se na známky, dokud uživatel nenapíše "konec". Ukládej je do
# seznamu, pak spočítej jejich počet a průměr.

znamky = []  # založíme složku na známky

while True: # nevím kolik inputů bude uživatel potřebovat
    vstup = input("Zadej známku nebo napiš konec:")
    if vstup == "konec":
        break
    znamky.append(int(vstup))

soucet = 0
for z in znamky:  # vím kolik známek zadal
    soucet += z
prumer = soucet/len(znamky)

print (f"Zadal jsi {len(znamky)} známek, průměr je {prumer:.1f}.")


```

```python
# Hádej číslo
# Program si náhodně zvolí číslo od 1 do 10. Uživatel hádá číslo, dokud ho neudhodne.
# Po každém pokusu program napoví: „Zkus větší číslo“ nebo „Zkus menší číslo“
# Když se trefí, program napíše: „Gratuluju! Uhodl jsi číslo


import random     # Načte modul, který umí generovat náhodná čísla.

tajne_cislo = random.randint(1, 10)  # Program si vybere náhodné číslo mezi 1 a 10 (včetně).
pokusy = 0  # Počítadlo, kolikrát uživatel hádal.

while True:  # Smyčka poběží, dokud ji ručně neukončíme (break).
    tip = int(input("Hádej číslo od 1 do 10: "))  # Uživatelský vstup, převedený na celé číslo.
    pokusy += 1

    if tip == tajne_cislo:
        print(f"Gratuluju! Uhodl jsi číslo {tajne_cislo} na {pokusy}. pokus.")
        break
    elif tip < tajne_cislo:
        print(f"Zadal jsi číslo {tip}. Zkus větší číslo.")
    else:
        print(f"Zadal jsi číslo {tip}. Zkus menší číslo.")
```

```python
# Ověření hesla
# Napište program, který: Má uložené správné heslo (např. "python123").
# Uživatel má tři pokusy, aby ho zadal správně.
# Pokud se mu to podaří, vypíše se: "Přístup povolen"
# Pokud třikrát zadá špatné heslo, vypíše se: "Přístup zamítnut"

spravne_heslo = "python123"
pokusy = 0

while pokusy <3:
    zadane_heslo = input("Zadej heslo:")

    if zadane_heslo == spravne_heslo:
        print("Přístup povolen")
        break
    else: 
        pokusy += 1
        print("Špatné heslo! Zbývá pokusů:", 3 - pokusy)

if pokusy == 3:
    print("Přístup zamítnut")
```

```python
# Horská dráha
# U vstupu na horskou dráhu stojí obsluha a kontroluje výšku návštěvníků.
# 1. Napiš funkci drah(vyska), která dostane výšku v centimetrech
#    a VRÁTÍ text:
#    - pokud je výška alespoň 120: "Nastupuj, jedeš!"
#    - jinak: "Ještě kousek povyrůst, zkus to příští rok."
#    (použij return, print uvnitř funkce nepatří)
# 2. Pomocí input se zeptej na výšku návštěvníka.
#    Pozor, input vrací text, výšku musíš převést na číslo: int(input(...))
# 3. Zavolej funkci s výškou a výsledek vytiskni.
#
# Ukázka 1:
# Kolik měříš (v cm)? 135
# Nastupuj, jedeš!
#
# Ukázka 2:
# Kolik měříš (v cm)? 110
# Ještě kousek povyrůst, zkus to příští rok.
#
# Bonus: přidej třetí možnost: kdo má přes 200 cm, dostane
# hlášku, že se do vozíku nevejde.

def drah(vyska):
    if vyska >= 120:
        return "Nastupuj, jedeš!"
    else:
        return "Ještě kousek povyrůst, zkus to příští rok."

vyska = int(input("Kolik měříš (v cm)? "))
zprava = drah(vyska)
print(zprava)
```

```python
# Pizza večer
#
# Objednáváš pizzy pro kamarády a chceš vědět, kolik to bude stát.
#
# 1. Napiš funkci doprava(soucet), která VRÁTÍ konečnou cenu:
#    - pokud je součet alespoň 500 Kč: doprava zdarma, vrátí se součet
#    - jinak se přičte 60 Kč za dopravu
#    (použij return, print uvnitř funkce nepatří)
#
# 2. Pomocí input se zeptej, kolik pizz objednáváš (int).
#
# 3. Pomocí cyklu for se u každé pizzy zeptej na její cenu
#    a ceny sčítej do proměnné soucet.
#    (nezapomeň, že soucet musí být před cyklem nastavený na 0)
#
# 4. Zavolej funkci se součtem a vytiskni, kolik zaplatíš.
#
# Ukázka:
# Kolik pizz objednáváš? 3
# Cena pizzy číslo 1: 189
# Cena pizzy číslo 2: 219
# Cena pizzy číslo 3: 169
# K zaplacení: 637 Kč (doprava zdarma)

# Vytvoření funkce, která sečte nákup a určí, jestli platíme dopravu
def doprava(soucet):
    if soucet >= 500:
        return soucet
    else:
        return soucet + 60

# Vložení hodnot
pocet = int(input("Kolik pizz objednáváš?"))
soucet = 0

# Cyklus vezme 1 pizzu, vezme cenu, a připoče další pizzu a další cenu
for pizza in range(pocet):
    cena = int(input(f"Cena pizzy číslo {pizza+1}"))
    soucet = soucet + cena

k_zaplaceni = doprava(soucet)
print(f"K zaplacení: {k_zaplaceni} Kč")
```

```python
# Skok do dálky
#
# Na školních závodech skáčou závodníci do dálky a ty zapisuješ výsledky.
# 1. Napiš funkci hodnoceni(nejdelsi), která dostane nejdelší skok v cm
#    a VRÁTÍ text:
#    - pokud je skok alespoň 450: "Rekord školy!"
#    - pokud je alespoň 400: "Hezký skok, jdeš do finále."
#    - jinak: "Trénuj dál, příště to vyjde."
#    (použij return, print uvnitř funkce nepatří)
# 2. Pomocí input se zeptej, kolik závodníků skákalo (int).
# 3. Pomocí cyklu for se u každého závodníka zeptej na délku jeho skoku (int)
#    a pamatuj si ten NEJDELŠÍ z nich v proměnné nejdelsi.
#    (nezapomeň, že nejdelsi musí být před cyklem nastavené na 0)
#
# 4. Zavolej funkci s nejdelším skokem a výsledek vytiskni.
#
# Ukázka:
# Kolik závodníků skákalo? 3
# Skok závodníka 1 (v cm): 412
# Skok závodníka 2 (v cm): 438
# Skok závodníka 3 (v cm): 425
# Nejdelší skok: 438 cm
# Hezký skok, jdeš do finále.

def hodnoceni(nejdelsi):
    if nejdelsi >= 450:
        return "Rekord školy!"
    elif nejdelsi >= 400:
        return "Hezký skok, jdeš do finále."
    else:
        return "Trénuj dál, příště to vyjde."

pocet = int(input("Kolik závodníků skákalo? "))
nejdelsi = 0
for i in range(pocet):
    skok = int(input(f"Skok závodníka {i + 1} (v cm): "))
    if skok > nejdelsi:
        nejdelsi = skok

print(f"Nejdelší skok: {nejdelsi} cm")
print(hodnoceni(nejdelsi))
```
