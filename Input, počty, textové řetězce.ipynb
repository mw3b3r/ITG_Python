```python
Přehled chyb v Pythonu

SyntaxError – Chybná gramatika kódu: 
Python nedokáže kód vůbec přečíst, protože porušuje jeho pravidla (např. chybějící závorka).

TypeError – Nesprávný datový typ: 
Například pokus o sčítání čísla a textu.

NameError – Neznámý název: 
Python nenašel proměnnou, protože ještě nebyla vytvořena nebo je špatně napsaná.

ZeroDivisionError – Dělení nulou: 
Matematická chyba – Python neumí dělit nulou.

ValueError – Neplatná hodnota: 
Hodnota neodpovídá očekávanému typu (např. snaha převést text na číslo, ale text neobsahuje čísla).

IndexError – Index mimo rozsah: 
Přístup k prvku seznamu, který neexistuje (např. požadavek na 5. prvek ze 3 položek).
```

```python
# Formátované řetězce (f-strings) v Pythonu

jméno = "Anna"
věk = 25

# klasický způsob spojování textu a proměnných

zpráva = "Ahoj, jmenuji se " + jméno + " a je mi " + str(věk) + " let."
print(zpráva)

zpráva = f"Ahoj, jmenuji se {jméno} a je mi {věk} let."
print(zpráva)
```

```python
# Sčítání a odčítání čísel zadaných uživatelem
# Zeptej se uživatele na dvě čísla. Spočítej jejich součet a rozdíl. Výsledek vypiš čitelně na obrazovku

# Načtení čísel od uživatele¨
cislo1 = float(input("Zadej první číslo: "))
cislo2 = float(input("Zadej druhé číslo: "))

#Výpočty
soucet = cislo1 + cislo2
rozdil = cislo1 - cislo2

#Výstup výsledků
print (f"Součet čísel {cislo1} a {cislo2} je: {soucet}") 
print (f"Rozdíl čísel {cislo1} a {cislo2} je: {rozdil}") 
```

```python
# Cvičení: Nákup. Zeptej se na cenu jednoho kusu a na počet kusů,
# spočítej celkovou cenu. Vypiš: "Zaplatíš 150 Kč."

cena = float(input("Zadej cenu jednoho kusu: "))
pocet = float(input("Počet kusů: "))
celkem = cena * pocet
celkem = round(celkem, 2)  # Zaokrouhlení na dvě desetinná místa
print (f"Zaplatíš {celkem} Kč.")
```

```python
# Cvičení: Průměr tří známek. Zeptej se na tři známky (celá čísla),
# spočítej průměr a zaokrouhli na jedno desetinné místo.
# Vypiš: "Tvůj průměr je 2.3"
znamka1 = int(input("Zadej první znamku: "))
znamka2 = int(input("Zadej druhou znamku: "))
znamka3 = int(input("Zadej třetí znamku: "))
prumer = (znamka1 + znamka2 + znamka3) / 3
prumer = round(prumer, 1)
print(f"Tvůj průměr je {prumer}")
```

```python
# Cvičení: Rozdělení účtu. Zeptej se na celkovou částku a počet lidí,
# spočítej, kolik platí každý, zaokrouhli na 2 desetinná místa.
# Vypiš: "Každý platí 216.67 Kč."

castka = float(input("Celková částka (Kč): "))
lidi = int(input("Počet lidí: "))
na_osobu = castka / lidi
na_osobu = round(na_osobu, 2)
print("Každý platí " + str(na_osobu) + " Kč.")
```

```python
# Cvičení (spojení všeho): Převod teploty z Celsia na Fahrenheit.
# Vzorec: F = C * 1.8 + 32
# Zeptej se na teplotu ve °C (může být desetinná), převeď a vypiš:
# "20.0 °C je 68.0 °F"

celsius = float(input("Teplota ve °C: "))
fahrenheit = celsius * 1.8 + 32
print(f"{celsius} °C je {fahrenheit} °F")
```

```python
# Oprava formátu jména
# Uživatel zadá své jméno malými písmeny a se zbytečnými mezerami na začátku a konci.
# Napiš program, který načte vstup od uživatele (např. " martina ").
# Odstraní přebytečné mezery na začátku a konci.
# Převede jméno na správný formát – první písmeno velké, ostatní malá.
# Vytiskne větu: Tvé jméno je Martina.

jmeno = input("Zadej své jméno: ")
# strip() odstraní mezery na začátku a konci
# capitalize() převede první písmeno na velké a ostatní na malá
opravene_jmeno = jmeno.strip().capitalize()
print(f"Tvé jméno je {opravene_jmeno}.")
```

```python
# Upper - převede všechna písmena na velká - upper()
# Lower - převede všechna písmena na malá - lower()
# replace() - nahradí zadaný text jiným textem - replace()
# find() - najde zadaný text a vrátí jeho pozici - find()
# split() - rozdělí text na seznam slov podle zadaného oddělovače - split()
# join() - spojí seznam slov do jednoho textu s oddělovačem - join()¨
# len() - vrátí počet znaků v řetězci - len()

```

```python
# replace() - nahradí zadaný text jiným textem - replace()
text = "Dneska je pátek a já chci už domů"
print(text.replace("pátek", "pondělí"))
```

```python
# find() - najde zadaný text a vrátí jeho pozici - find()
# Bacha - Python počítá indexy od nuly, takže první znak má index 0, druhý znak má index 1 atd.

text = "Dnesk a je pátek a já chci už domů"
print(text.find("a já"))  # Vrátí index, kde se nachází písmeno "a"
```

```python
# split() - rozdělí text na seznam slov podle zadaného oddělovače - split()
text = "Dneska je pátek a já chci už domů"
print(text.split(" "))  # Rozdělí text na seznam slov podle mezer   
```

```python
# split() - rozdělí text na seznam slov podle zadaného oddělovače - split()
text = "Dneska je pátek a já chci už domů. Protože jedeme na chatu a tam je super."
print(text.split(". "))  # Rozdělí text na seznam slov podle mezer   
```

```python
# join() - spojí seznam slov do jednoho textu s oddělovačem - join()
slova = ['Dneska', 'je', 'pátek', 'a', 'já', 'chci', 'už', 'domů']
print(" ".join(slova))  # Spojí seznam slov do jednoho textu s oddělovačem tečka'
# syntaxe: ("spojovatel". příkaz(proměnná))
```

```python
# Uprav vstup a slož větu. Uživatel může napsat mezery a malá
# písmena (třeba "  brno "). Ořízni mezery, dej velké první písmeno,
# vypiš: "Bydlíš ve městě Brno."


mesto = input("Zadej město: ")
mesto = mesto.strip().capitalize()
print(f"Bydlíš ve městě {mesto}.")
```
