```python
# procvičení probrané látky:

# Úroveň 1: Zeptej se na věk v letech. Spočítej, kolik je to zhruba
# dní, hodin a minut (rok ber jako 365 dní). Vypiš třemi větami.

# Úroveň 2: Zeptej se, kolik hodin denně spíš (může být 7.5).
# Spočítej, kolik let svého života jsi prospal/a. Zaokrouhli na 1 místo.

# Úroveň 3: Změř si puls: 15 sekund počítej údery, výsledek zadej.
# Spočítej údery za minutu a kolikrát ti srdce už bouchlo za celý život.

# Úroveň 4: Zeptej se, kolik hodin denně jsi na mobilu a od kolika let
# ho máš. Kolik let života už jsi strávil/a na mobilu?
# Je to víc než rok? (odpověď True/False)
```

```python
# Úroveň 1: Zeptej se na věk v letech. Spočítej, kolik je to zhruba
# dní, hodin a minut (rok ber jako 365 dní). Vypiš třemi větami.

vek = int(input("Kolik ti je let? "))

dny = vek * 365
hodiny = dny * 24
minuty = hodiny * 60

print(f"Žiješ zhruba {dny} dní.")
print(f"To je {hodiny} hodin.")
print(f"A {minuty} minut.")
```

```python
# Úroveň 2: Zeptej se, kolik hodin denně spíš (může být 7.5).
# Spočítej, kolik let svého života jsi prospal/a. Zaokrouhli na 1 místo.

vek = int(input("Kolik ti je let? "))
spanek = float(input("Kolik hodin denně spíš? "))

prospano_let = vek * spanek / 24

print(f"Prospal/a jsi zhruba {prospano_let:.1f} let.")
```

```python
# Úroveň 3: Změř si puls: 15 sekund počítej údery, výsledek zadej.
# Spočítej údery za minutu a kolikrát ti srdce už bouchlo za celý život.

vek = int(input("Kolik ti je let? "))
tepy_15s = int(input("Kolik úderů jsi napočítal/a za 15 sekund? "))

tepy_minuta = tepy_15s * 4
tepy_zivot = tepy_minuta * 60 * 24 * 365 * vek

print(f"Tvůj puls je {tepy_minuta} úderů za minutu.")
print(f"Srdce ti bouchlo zhruba {tepy_zivot} krát.")
```

```python
# Úroveň 4: Zeptej se, kolik hodin denně jsi na mobilu a od kolika let
# ho máš. Kolik let života už jsi strávil/a na mobilu?
# Je to víc než rok? (odpověď True/False)

vek = int(input("Kolik ti je let? "))
mobil_hodin = float(input("Kolik hodin denně jsi na mobilu? "))
mobil_od = int(input("Od kolika let máš mobil? "))

mobil_let = (vek - mobil_od) * mobil_hodin / 24

print(f"Na mobilu jsi strávil/a zhruba {mobil_let:.1f} let.")
print("Je to víc než rok?", mobil_let > 1)
```

```python
# Podmínky a větvení
# ------------------------------------------------------------

#Máte proměnnou teplota, která představuje aktuální teplotu ve stupních Celsia.
# Podle její hodnoty vypište odpovídající text:
# pokud je více než 25 °C, vypiš: Je horko
# pokud je mezi 15 a 25 °C včetně, vypiš: Příjemné počasí
# pokud je méně než 15 °C, vypiš: Je zima

teplota = 20

if teplota > 25:
    print("Je horko")
elif teplota >= 15 and teplota <= 25:
    print("Příjemné počasí")
else:
    print("Je zima")
```

```python
# Cvičení: Vstupné do zoo podle věku. Do 6 let zdarma, do 18 dětské 80 Kč,
# od 65 seniorské 100 Kč, jinak plné 150 Kč.

vek = int(input("Kolik ti je let? "))

if vek < 6:
    print("Vstup zdarma")
elif vek < 18:
    print("Dětské vstupné: 80 Kč")
elif vek >= 65:
    print("Seniorské vstupné: 100 Kč")
else:
    print("Plné vstupné: 150 Kč")
```

```python
# NOVÉ: podmínka může být uvnitř jiné podmínky (vnořené if).
# Vnitřní if se řeší, jen když platí ten vnější.

# Cvičení: Přihlášení. Nejdřív ověř přihlašovací jméno "admin".
# Když sedí, teprve pak zkontroluj heslo "tajne". Rozliš tři případy:
# špatné jméno, správné jméno ale špatné heslo, obojí správně.

jmeno = input("Jméno: ")
heslo = input("Heslo: ")

if jmeno == "admin":
    if heslo == "tajne":
        print("Přihlášení proběhlo úspěšně")
    else:
        print("Špatné heslo")
else:
    print("Neznámé jméno")
```

```python
# Cvičení: Bankomat. Na účtu je 500 Kč, správný PIN je "1234".
# Zeptej se na PIN. Když nesedí, vypiš chybu. Když sedí, teprve pak
# se zeptej, kolik chce uživatel vybrat. Když je na účtu dost peněz,
# vydej je a vypiš zůstatek. Když ne, vypiš, že na účtu není dost peněz.

zustatek = 1000
pin = input("zadej PIN")

if pin == "1234":
    castka = int(input("Jakou částku chceš vybrat?"))
    if castka <= zustatek:
        zustatek = zustatek - castka
        print(f"Vydávám {castka} Kč. Na účtu zbývá {zustatek} Kč.")
    else:
        print("Na účtu nemáš dost peněz")
else:
    print("špatný PIN, zkus to znova")
```

```python
# Cvičení: Kino. Zeptej se, jestli má divák lístek (ano/ne).
# Když lístek nemá, pošli ho k pokladně. Když ho má, teprve pak
# zkontroluj věk: film je od 15 let. Rozliš tři případy:
# nemá lístek, má lístek ale je mu méně než 15, má lístek a je mu 15+.

listek = input("Máš lístek? (ano/ne): ").strip().lower()

if listek == "ano":
    vek = int(input("Kolik ti je let? "))
    if vek >= 15:
        print("Pojď dál, příjemnou zábavu!")
    else:
        print("Promiň, tenhle film je od 15 let.")
else:
    print("Nejdřív si kup lístek u pokladny.")
```

```python
# Cvičení: Párty. Máš seznam pozvaných hostů. Zeptej se na jméno.
# Když host na seznamu není, nepusť ho dovnitř. Když na seznamu je,
# teprve pak se zeptej na věk: od 18 let dostane na baru welcome drink,
# mladší dostane limonádu. Rozliš tři případy.


hoste = ["Anna", "Petr", "Eva", "Tomáš"]
jmeno = input("Jak se jmenuješ? ").strip().capitalize()

if jmeno in hoste:
    vek = int(input("Kolik ti je let? "))
    if vek >= 18:
        print(f"Vítej, {jmeno}! Na baru na tebe čeká welcome drink.")
    else:
        print(f"Vítej, {jmeno}! Na baru na tebe čeká limonáda.")
else:
    print("Promiň, nejsi na seznamu hostů.")
```

```python
# Smyčky (cykly)
# Napište program, který spočítá součet všech čísel 
# od 1 do 100 pomocí smyčky for.

soucet = 0   # nastavení výchozího stavu

for cislo in range(1, 101):
    soucet += cislo

print("Součet čísel od 1 do 100 je:", soucet)
```

```python
# Napište rituál programátora, který bez pěti káv nezačne den. 
# Smyčkou mu spočtěte kávy tak, aby mohl začít pracovat

for kafe in range(1,6):
    print(f"hrnek kafe č. {kafe}")
print("Mazej makat a nepij furt to kafe!")
```

```python
# NOVÉ: for umí projít přímo prvky seznamu, ne jen čísla z range().
# for jidlo in seznam: vezme postupně každý prvek a uloží ho do jidlo.

# Cvičení: Máš seznam oblíbených jídel. Vypiš každé na vlastní řádek.

jidla = ["pizza", "knedliky", "dort", "stejk", "nanuk"]
for jidlo in jidla:  # nepoužívam range, hledám v seznamu a tam se používá in
    print(jidlo)
```

```python
# Cvičení: Máš seznam jmen. U každého vypiš, že je na seznamu.
# Výstup: "Anna je na seznamu." atd.

jmena = ["Anna", "Petr", "Eva"]
for jmeno in jmena:
    print(f"{jmeno} je na seznamu.")
```

```python
# Cvičení: Spočítej, kolikrát je ve slově písmeno "a".
# Založ počítadlo na 0, projdi písmena a u každého "a" přičti 1.

slovo = input("Zadej slovo:")
soucet = 0

for pismeno in slovo.lower():
    if pismeno == "a":
        soucet += 1
print(f"Pismeno a je ve slově {soucet} krát")
```

```python
# Cvičení: Součet a průměr seznamu čísel.
# (Stejný trik jako součet 1 až 100, jen teď sčítáš prvky seznamu.)

teploty = [21, 19, 24, 17, 22]
soucet = 0

for t in teploty:
    soucet += t
prumer = soucet /len(teploty)

print(f"Průměrná teplota je {prumer:.1f} °C.")
```

```python
# Cvičení: Kolik lidí v seznamu je dospělých (18+)?
# Projdi věky a u každého, kdo má 18 a víc, přičti k počítadlu.

veky = [15, 22, 17, 30, 12, 45]
dospeli = 0

for v in veky:
    if v >= 18:
        dospeli += 1
print(f"Dospělých je {dospeli}")
```

```python
# Založ prázdný seznam [], v cyklu do něj přidávej jen vybrané prvky.
# Cvičení: Ze seznamu čísel vyber jen sudá do nového seznamu.

cisla = [3, 8, 15, 6, 21, 10]
suda = []

for c in cisla:
    if c % 2 == 0:    # pokud má císlo po dělení dvěma za des. čárkou nulu, 
        suda.append(c)  # pridá se tohle číslo do seznamu "suda"
print("Sudá čísla:", suda)

```
