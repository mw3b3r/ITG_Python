```python
# Práce  s textovými řetězci

# Upper - převede všechna písmena na velká - upper()
# Lower - převede všechna písmena na malá - lower()
# replace() - nahradí zadaný text jiným textem - replace()
# find() - najde zadaný text a vrátí jeho pozici - find()
# split() - rozdělí text na seznam slov podle zadaného oddělovače - split()
# join() - spojí seznam slov do jednoho textu s oddělovačem - join()
```

```python
# Hranaté závorky a Indexing - umožňují přístup k jednotlivým znakům v textu podle jejich pozice
# Počítá se od nuly, protože Python bere první znak jako 0
# Iniciály. Z Jana Nováka uděláme J.N.

jmeno = input("Jméno:")
prijmeni = input("Příjmení:")
inicialy = jmeno[0].upper() +"."+ prijmeni[0].upper() + "."
print(f" Tvoje inciály jsou: {inicialy}")
```

```python
# Výřez několika písmen z textu - Slicing
# Tvoří se [: číslo] nebo [číslo :]
# text[2:] - vezme všechno od třetího písmene dál
# text[:2] - vezme všechno do třetího písmene
# text[3:6] - vezme písmena od čtvrtého do šestého

jmeno = input ("jmeno:")
cislo = int(input("cislo:"))    
# vytvořit přezdívku z jména a číla. tři písmena ze jména + číslo
prezdivka = jmeno[3:6] + str(cislo)
print(f"Tvoje přezdívka je:{prezdivka}")
```

```python
# Cvičení: Počet slov. Zeptej se na větu, rozděl ji na slova přes .split()
# a spočítej je pomocí len(). Vypiš: "Tvá věta má 5 slov."

veta = input("Zadej vetu:")
slova = veta.split(" ")
pocet = len(slova)
print(f"Tvá věta má {pocet} slov.")
```

```python
#f-string umí rovnou zaokrouhlit. {cislo:.1f} zobrazí jedno
# desetinné místo, {cislo:.2f} dvě. Nahradí to round() + str().

vyska = float(input("Zadej výšku v metrech. Jako oddělovač použij tečku."))
hmotnost = float(input("Zadej hmotnost v kilogramech. Jako oddělovač použij tečku."))
bmi = hmotnost / (vyska * vyska)
print(f"Tvoje BMI je: {bmi:.2f}")
```

```python
# Cvičení (spojení všeho): Generátor přihlašovacího jména.
# Zeptej se na příjmení a rok narození. Login = první tři písmena
# příjmení malými + poslední dvě číslice roku.
# Z, "Novák", "2008" vznikne "nov08".

prijmeni = input("Příjmení: ")
rok = input("Rok narození: ")
login = prijmeni[:3].lower() + rok[2:]
print(f"Tvůj login je {login}")
```

```python
# Porovnávání a logika

# Porovnej dva věky
# Zeptej se na věk dvou osob. Použij porovnání podle operátorů ==, >, < , <=, >=

vek_a = int(input("Zadej věk osoby A: "))
vek_b = int(input("Zadej věk osoby B: "))

# Jsou oba stejně staří? 
print("Jsou oba stejně staří?", vek_a == vek_b)
# Je A straší než B? 
print("Je A starší než B?", vek_a > vek_b)
# Je B mladší než 18 let? 
print("Je B mladší než 18 let?", vek_b <18)
# Jsou ona dospělí? (18 a více)
print("Jsou ona dospělí?", vek_b >= 18 and vek_a >=18)
```

```python
# Porovnej dvě zadaná jména a zjisti:
# Jsou stejná? 
# Mají stejný počet písmen? 
# Je první jméno „větší“ než druhé podle abecedy?
# Obsahuje některé jméno písmeno „a“?

jmeno1 = input("Zadej první jméno:")
jmeno2 = input("Zadej druhé jméno:")

# Jsou stejná? 
print("Jsou jména stejná?", jmeno1==jmeno2)
# Mají stejný počet písmen? 
print("Jsou jména stejně dlouhá?", len(jmeno1) == len(jmeno2)) 
# Je první jméno „větší“ než druhé podle abecedy?
print ("Je první jméno „větší“ než druhé podle abecedy?",jmeno1 > jmeno2)

# POZOR NOVINKA
# Obsahuje některé jméno písmeno „a“?
obsahuje_a = ("a" in jmeno1) or ("a" in jmeno2)
print("Obsahuje některé jméno písmeno „a“?", obsahuje_a)
```

```python
# NOVÉ: != znamená "nerovná se". Vrací True, když jsou hodnoty různé.
# Cvičení: Zeptej se na dvě čísla a zjisti, jestli jsou různá.

a = int(input("První číslo: "))
b = int(input("Druhé číslo: "))
print("Jsou čísla různá?", a != b)
print("Jsou čísla stejná?", a == b)
```

```python
# Cvičení: Výsledek porovnání si můžu uložit do proměnné a pak s ním
# dál pracovat. Zjisti, jestli je číslo v rozsahu 1 až 10 (včetně).
# Pomůcka: musí platit dvě věci naráz (>= 1 a zároveň <= 10).

cislo = int(input("Zadej číslo:"))
v_rozsahu = cislo >= 1 and cislo <=10
print("je číslo v rozmězí 1 a 10?", v_rozsahu)
```

```python
# Cvičení: Oprava hledání písmene z minula. Zjisti, jestli jméno
# obsahuje "a" bez ohledu na velikost písmen (A i a).

jmeno = input("Zadej jméno:")
obsahuje_a = "a" in jmeno.lower()
print ("Obsahuje písmeno A nebo a?", obsahuje_a)
```

```python
# Kontrola přezdívky pro registraci.
# Přezdívka je v pořádku, když má víc než 3 znaky A ZÁROVEŇ
# se nerovná zakázanému slovu "admin".

prezdivka = input("Zadej přezdívku:")
# proměnná na to, jestli je přezdívka dost dlouhá
dost_dlouha = len(prezdivka) > 3
# proměnná na zakázané slovo admin
admin = prezdivka!= "admin"
# proměnná kde to spojím
prezdivka_ok = dost_dlouha and admin
print("Je přezdívka ok?", prezdivka_ok)
```

```python
# Práce se seznamy

# Uživatel zadá 3 svá oblíbená jídla. Tvůj program je uloží do seznamu a vypíše zpět.
# Požádej uživatele, aby postupně zadal 3 jídla (např. input).
# Ulož je do seznamu. Vypiš celý seznam. Vypiš první jídlo ze seznamu zvlášť.

# uživatel zadá tři svoje oblíbená jídla
jidlo1 = input("Zadej první jídlo:")
jidlo2 = input("Zadej druhé jídlo:")
jidlo3 = input("Zadej třetí jídlo:")

# vytvořím seznam
oblibena_jidla = [jidlo1, jidlo2, jidlo3]

# vypíšu celý seznam
print("Tvoje oblíbená jídla jsou:", oblibena_jidla)

# Vypíšu jenom první jídlo
print("Tvoje nejoblíbenější jídlo je:", oblibena_jidla[0])
```

```python
# Nákupní seznam
# Začni s malým seznamem nákupních položek a nech uživatele přidat další položku.
# Zadání: Vytvoř seznam s několika položkami, např. ["mléko", "vejce", "chléb"].
# Požádej uživatele o další položku. Přidej ji do seznamu.
# Vypiš aktuální nákupní seznam.

nakup = ["mléko", "chleba", "vajíčka"]
print(nakup)

# Přidání položky do seznamu
nova_polozka = input("Co chceš přidat na seznam?")
nakup.append(nova_polozka)
print(nakup)

# Odebrání položky ze seznamu
odebrana_polozka = input("Co chceš vyřadit?")
nakup.remove(odebrana_polozka)
print(nakup)

```

```python
# Cvičení: Máš seznam čekajících ve frontě. Vypiš, kdo je první na řadě
# a kdo je úplně poslední.

fronta = ["Anna", "Pepa", "Franta", "Tonda"]
print(fronta[0])  # Kdo je ve frontě první
print(fronta[-1])  # Kdo je ve frontě poslední
```

```python
# Vkládání do seznamu

fronta = ["Anna", "Pepa", "Franta", "Tonda"]

#Vložení přes append - vloží novou položku na konec seznamu
nove_jmeno = input("Co chceš přidat na seznam?")
fronta.append(nove_jmeno)

# Vložení přes insert - vloží položku na konkrétní místo v seznamu
nove_jmeno2 = input("Koho chceš ještě na seznam?")
fronta.insert(0, nove_jmeno2 )


print(fronta)
```

```python
# Jak něco v seznamuj najít (IN)

fronta = ["Anna", "Pepa", "Franta", "Tonda"]
jmeno = input ("Koho chceš zkontrolovat? ")
print ("Je na seznamu?", jmeno in fronta)
```

```python
# Jak seřadit seznam podle abecedy (sorted)

fronta = ["Anna", "Pepa", "Franta", "Tonda"]
serazena = sorted(fronta)
print("podle abecedy:", serazena)

```

```python
# Cvičení (spojení všeho): Správa čekací listiny.
# Začni se seznamem tří lidí. Přidej nového na konec (append),
# jednoho odeber (remove), vypiš aktuální listinu i její délku (len)
# a nakonec ji vypiš seřazenou (sorted).

listina = ["Anna", "Petr", "Eva"]
novy = input("Kdo se přihlásil? ")
listina.append(novy)
odchazi = input("Kdo se odhlásil? ")
listina.remove(odchazi)
print("Aktuální listina:", listina)
print(f"Počet lidí: {len(listina)}")
print("Seřazeně:", sorted(listina))
```

```python
# Varianta 2
# Řešení bez INPUT

listina = ["Anna", "Petr", "Eva"]
listina.append("Jana")
listina.remove("Petr")
print("Aktuální listina:", listina)
print("Počet lidí:", len(listina))
print("Seřazeně:", sorted(listina))
```
