```python
# Slovníky a n-tice
# -----------------------------------------------------------------

# Vytvoř slovník, kde: klíčem bude název knihy, hodnotou bude jméno autora
# Nechte uživatele zadat název knihy (input()).
# Program vypíše autora, pokud knihu ve slovníku najde. 
# Pokud tam kniha není, vypíše zprávu „Knihu nemáme v databázi.“

# vytvoříme slovník
knihy = {
    "Lovci mamutů" : "Eduard Štorch",
    "Babička":"Božena Němcová",
    "Hobit":"J.R.R. Tolkien",
    "Harry Potter":"J.K.Rowling"
}
# Název knihy je klíč, autor je hodnota

nazev = input("Jakou knihu hledáš?")
# uživatel zadá název knihy

autor = knihy.get(nazev)
# sáheneme do slovníku knihy přes get a čekneme název co zadal uživatel
# Uložíme to do proměnné autor

if autor:
    print(f"Autorem knihy {nazev} je {autor}")
else:
    print("Knihu nemáme v databázi.")

```

```python
# Přidat knížku do slovníku

# slovnik[klic] = hodnota
knihy["Pýcha a předsudek"] = "Jane Austen"
print(knihy)
```

```python
# Odstranit knížku ze slovníku

# del slovník[klíč]
del knihy["Babička"]
print(knihy)
```

```python
# Přepsání klíče je problém, protože: 

knihy["Pán prstenů"] = "J.R.R. Tolkien"
print(knihy)

# Pokud se klíč 100% neshoduje s už uloženým, přidá se jako nová hodnota. 
```

```python
# Přepsání hodnoty je v pohodě:

knihy["Pýcha a předsudek"] = "Jane Austenová"
print(knihy)
```

```python
# Vypiš všechny knihy

for nazev in knihy.keys():
    print(nazev)
```

```python
# Vypiš všechny autory

for autor in knihy.values():   # pro každý název v slovníky knihy najdi hodnotu
    print(autor)
```

```python
# Vypiš všechno

for autor, nazev in knihy.items():  
    print(f"{autor} - {nazev}")
```
